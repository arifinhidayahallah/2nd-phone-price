```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

phones_df = pd.read_csv('datasets/extracted_features_data/phones.csv')
```

```python
categorical_cols = phones_df.select_dtypes(include=['string']).columns
numerical_cols = phones_df.select_dtypes(include=['number']).columns
encoder = OrdinalEncoder(
    handle_unknown='use_encoded_value',
    unknown_value=-1
)

encoded_phones_df = phones_df.copy()
encoded_phones_df[categorical_cols] = encoder.fit_transform(phones_df[categorical_cols])
encoded_phones_df[categorical_cols].columns
```

```python
encoded_phones_df['brand'].value_counts()
```

```python
decoded = encoder.inverse_transform(encoded_phones_df[categorical_cols])
```

```python
comparison = pd.DataFrame({
    'encoded':  encoded_phones_df['brand'],
    'decoded': decoded[:,list(categorical_cols).index('brand')]
})
comparison.value_counts().sort_index()
```

```python
import joblib

joblib.dump(encoder,"models/features_encoder.joblib")
```

```python
encoded_phones_df.to_csv('datasets/transformed_data/phones.csv', index=False)
encoded_phones_df.to_json('datasets/transformed_data/phones.json', index=False)
pd.DataFrame({'features': categorical_cols}).to_csv('datasets/transformed_data/categorical_features.csv', index=False)
pd.DataFrame({'features': numerical_cols}).to_csv('datasets/transformed_data/numerical_features.csv', index=False)
pd.DataFrame({'features': encoded_phones_df.columns}).to_csv('datasets/transformed_data/selected_features.csv', index=False)
```
