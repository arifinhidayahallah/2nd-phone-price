```python
null_features = phones_df.isna().sum().sort_values(ascending=False)
# null_features

# for i in null_features[null_features == phones_len].index:
#     print(i)
# null_features

features = ['location','rating','sold','candidate_model','device_type','aspect_ratio', 
     'price_old', 'price_savings','price_official', 'price_unofficial','price_updated',
     'price_variants','ruggedness','ip_rating','user_interface', 'fabrication', 
     'status','image_url','brightness', 'colors', 'scraped_at', 'detail_url', 'model']

for k, v in null_features.items():
    # print(k)
    if k in features:
        print(k, v)
        
```

```python
phones_df.drop(columns=null_features[null_features == phones_len].index.union(
    ['location','rating','sold','candidate_model','device_type','aspect_ratio', 
     'price_old', 'price_savings','price_official', 'price_unofficial','price_updated',
     'price_variants','ruggedness','ip_rating','user_interface', 'fabrication', 
     'status','image_url','brightness', 'colors', 'scraped_at', 'detail_url', 'model'
    ]), inplace=True)
```

```python
phones_df['volte'].value_counts()
```

```python
phones_df.isna().sum().sort_values(ascending=False )
```

```python
import numpy as np


phones_df.fillna({"waterproof":"No", "usb_type_c":"No", "ram_type":"Undefined", "storage_type":"Undefined", "quick_charging":"No","face_unlock":"No", "usb_otg": "No", "refresh_rate":"60 Hz", "screen_protection":"Undefined", "notch":"No", "audio_jack": "No", "expandable_memory": "No"}, inplace=True)

phones_df.loc[phones_df['bluetooth'] == 'Yes', 'bluetooth'] = "v5.0"
phones_df.loc[phones_df['volte'] == 'VoLTE', 'volte'] = "Yes"
phones_df.loc[phones_df['release_date'].str.lower() == 'not announced yet', 'release_date'] = np.nan
phones_df.loc[phones_df['release_date'].str.lower().str.contains('bangladesh'), 'release_date'] = phones_df['release_date'].str.replace('(Bangladesh)','')

phones_df['price'] = phones_df['price'].str.replace('Rp','').str.replace('.','').astype(float)
phones_df['os_version'] = phones_df['os_version'].str.replace('v','')
phones_df['audio_jack'] = phones_df['audio_jack'].str.replace(' ','')
phones_df['release_date'] = phones_df['release_date'].str.replace('Exp. ','')
```

```python
phones_df['thickness'] = (phones_df['thickness'].str.extract(r"(\d+(?:\.\d+)?)").astype(float))
phones_df['screen_to_body_ratio'] = (phones_df['screen_to_body_ratio'].str.extract(r"(\d+(?:\.\d+)?)").astype(float))
phones_df['width'] = (phones_df['width'].str.extract(r"(\d+(?:\.\d+)?)").astype(float))
phones_df['height'] = (phones_df['height'].str.extract(r"(\d+(?:\.\d+)?)").astype(float))
phones_df['weight'] = (phones_df['weight'].str.extract(r"(\d+(?:\.\d+)?)").astype(float))

phones_df['weight'] = phones_df['weight'].fillna(
    phones_df['weight'].median()
)
phones_df['thickness'] = phones_df['thickness'].fillna(
    phones_df['thickness'].median()
)
phones_df['screen_to_body_ratio'] = phones_df['screen_to_body_ratio'].fillna(
    phones_df['screen_to_body_ratio'].median()
)
phones_df['width'] = phones_df['width'].fillna(
    phones_df['width'].median()
)
phones_df['height'] = phones_df['height'].fillna(
    phones_df['height'].median()
)
```

```python
phones_df.dropna(inplace=True)
# phones_df.isna().sum().sort_values(ascending=False)
```

```python

phones_df['quick_charging'] = phones_df['quick_charging'].str.extract(r"(\d+\s*W|No)")
phones_df['refresh_rate'] = phones_df['refresh_rate'].str.extract(r"(\d+)").astype(int)
phones_df['battery_capacity'] = phones_df['battery_capacity'].str.findall(r"(\d+)").str[0].astype(float)
phones_df['screen_size'] = phones_df['screen_size'].str.findall(r"(\d+\s*x\s*\d+)").str[0].astype(str)
phones_df['cpu_cores'] = phones_df['cpu_cores'].str.findall(r"(\d+)").str[0].astype(int)
phones_df['pixel_density'] = phones_df['pixel_density'].str.findall(r"(\d+)").str[0].astype(int)
phones_df['architecture'] = phones_df['architecture'].str.findall(r"(\d+)").str[0].astype(int)
```
