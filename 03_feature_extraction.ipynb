```python
import re

def extract_ram_storage(title):
    
    possible_formats = [
        r'(\d+)\s*gb(\d+)\s*gb',
        r'(\d+)gb\s*(\d+)gb\s*',
        r'(\d+)\s*/\s*(\d+)',
        r'(\d+)\s*/\s*(\d+)\s*gb',
        r'(\d+)\s*gb\s*/\s*(\d+)\s*gb',
        r'ram\s*(\d+)\s*(\d+)\s*gb',
        r'ram\s*(\d+)\s+rom\s*(\d+)\s*gb',
        r'ram\s*(\d+)\s*gb\s+rom\s*(\d+)\s*gb'
    ]
    
    if pd.isna(title):
        return pd.Series([None, None])
    
    title = title.lower()
    
    for format in possible_formats:
        match = re.search(format, title)

        if match:
            return pd.Series([match[1], match[2]])
    
    
    return pd.Series([None, None])
```

```python
phones_df[['ram','internal_storage']] = phones_df['title'].apply(extract_ram_storage)
```

```python
from datetime import datetime

def reformat_date(str_date):
    date_formats = [
        '%d%B%Y',
        '%d%b%Y',
        '%d,%B,%Y',
        '%d,%b,%Y',
    ]
    
    for format in date_formats:
        try:
            parsed =  datetime.strptime(str_date.replace(" ",""), format)
            return parsed
        except:
            continue

phones_df['release_date'] = phones_df['release_date'].apply(lambda x: reformat_date(x))

phones_df['released_year'] = phones_df['release_date'].dt.year
phones_df['released_month'] = phones_df['release_date'].dt.month
phones_df['released_day'] = phones_df['release_date'].dt.day
        
```
