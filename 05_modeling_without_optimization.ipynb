```python
import pandas as pd
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics  import  (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    root_mean_squared_error
)
```

```python
phones_df = pd.read_csv('datasets/transformed_data/phones.csv')
```

```python
split_combination = [.2, .1]
```

```python
X_data = phones_df.drop(columns=['price'])
y_data = phones_df['price']
```

```python
for i in split_combination:
    X, X_test, y, y_test = train_test_split(
    X_data,
    y_data,
    test_size=i,
    random_state=42
    )

    N_FEATURES = len(X.columns)

    model = RandomForestRegressor(
        random_state=42
    )

    model.fit(X, y)
    y_pred = model.predict(X_test)


    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = root_mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    print(f"Split Combination : {i}")
    print(f"MAE: {mae}")
    print(f"MSE: {mse}")
    print(f"RMSE: {rmse}")
    print(f"R2 Score: {r2}")
    print(f"-----------------------------------")
```

```python
importance = pd.DataFrame({
    "feature": X.columns,
    "importance": model.feature_importances_
})

importance = importance.sort_values(
    by="importance",
    ascending=False
)

print(importance)
```

```python
import joblib

joblib.dump(model, "models/2nd_phone_price_without_optimization.pkl")
```
